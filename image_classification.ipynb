```python
import tensorflow as tf
import matplotlib.pyplot as plt
from  tensorflow.keras import layers
```

```python
(x_train,y_train),(x_test,y_test)= tf. keras.datasets.fashion_mnist.load_data()
```

```python
print("length of train data",len(x_train))
```

```python
print("length of test data",len(y_test))
```

```python
print("size of image",x_train[0].shape)
```

```python
plt.imshow(x_train[0])
plt.grid()
plt.show()
```

```python
X_train= x_train/28
X_test= x_test/28
```

```python
X_train= X_train.reshape(-1,28,28,1)
X_test= X_test.reshape(-1,28,28,1)
```

```python
X_test
```

```python
#create input layers
input_layer= layers.Input(shape=(28,28,1))

#CNN model building
model= tf.keras.Sequential([input_layer,
                            layers.Conv2D(32,(3,3),activation='relu'),
                            layers.Conv2D(32,(3,3),activation='relu'),
                            layers.Conv2D(32,(3,3),activation='relu'),
                            layers.MaxPool2D(),
                            layers.Flatten(),
                            layers.Dense(10,activation='softmax')])
```

```python
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
model.fit(X_train,y_train,epochs=10)
```

```python
loss, accuracy=model.evaluate(X_train,y_train)
print("test accuracy:", accuracy)
print("error:", loss)
```

```python
loss, accuracy=model.evaluate(X_test,y_test)
print("test accuracy:", accuracy)
print("error:", loss)
```

```python
model.summary()
```

```python
y_pred= model.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix, classification_report, ConfusionMatrixDisplay
import numpy as np
```

```python
y_pred_labels = np.argmax(y_pred, axis=1)
if y_test.ndim > 1:
    y_test_labels = np.argmax(y_test, axis=1)
else:
    y_test_labels = y_test

confusion_matrix(y_pred_labels,y_test_labels)
```

```python
print(classification_report(y_pred_labels,y_test_labels))
```

```python
ConfusionMatrixDisplay.from_predictions(y_test_labels,y_pred_labels)
```
